```python
import io
import os
import pandas as pd
import paramiko
import matplotlib.pyplot as plt
df=pd.read_pickle('combined_data.pkl')
```

```python
alt = [40, 60, 80, 100, 120, 140, 160, 180, 200, 220, 240, 260, 280, 300]
means = []

# 各高度の平均値を計算してリストに溜める
for h in alt:
    mean_val = df[f"{h}m Wind Speed (m/s)"].mean()
    means.append(mean_val)

# 横軸：平均風速、縦軸：高度 でプロット
#plt.figure(figsize=(6, 8))
plt.plot(means, alt, marker="o", color="blue", linewidth=2)
#plt.xlabel("Mean Wind Speed (m/s)")
#plt.ylabel("Altitude (m)")
plt.grid(True)
plt.show()
```

```python
alt = [40, 60, 80, 100, 120, 140, 160, 180, 200, 220, 240, 260, 280, 300]

plt.figure(figsize=(12, 6))

for h in alt:
    dir=((df[f"{h}m Wind Direction (ｰ)"] + 180) % 360) - 180
    # .mean() を外すことで、X軸と同じ4002行のデータになり、サイズが一致します
    plt.plot(
        df["Timestamp (end of interval)"],
        dir,
        label=f"{h}m",
        alpha=0.7,  # 線が重なって見にくくなるのを防ぐため少し透過
    )
plt.ylim(-180, 180)
plt.yticks([-180, -90, 0, 90, 180])
plt.xlabel("Timestamp")
plt.ylabel("Wind Direction")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")  # 凡例を外側に配置
plt.grid(True)
plt.tight_layout()
plt.show()
```
